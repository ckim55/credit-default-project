# 01 EDA: Credit Card Default Dataset

This notebook loads the UCI credit default dataset and performs basic exploratory analysis. Run this notebook before modeling to understand class imbalance, feature distributions, and relationships between repayment behavior and default risk.

```python
import sys
from pathlib import Path

ROOT = Path.cwd().parent
if str(ROOT) not in sys.path:
    sys.path.append(str(ROOT))

import pandas as pd
import matplotlib.pyplot as plt

from src.data import load_local_or_download, split_X_y
from src.features import add_engineered_features
```

```python
df = load_local_or_download(ROOT / "data/raw/credit_default_raw.csv")
df.head()
```

```python
df.shape, df.columns.tolist()
```

```python
df.isna().sum().sort_values(ascending=False).head(10)
```

```python
target_col = "default_next_month"
class_counts = df[target_col].value_counts().sort_index()
class_rates = df[target_col].value_counts(normalize=True).sort_index()
pd.DataFrame({"count": class_counts, "rate": class_rates})
```

```python
plt.figure(figsize=(5, 4))
class_counts.plot(kind="bar")
plt.title("Target Class Distribution")
plt.xlabel("Default Next Month")
plt.ylabel("Count")
plt.tight_layout()
```

```python
X_raw, y = split_X_y(df)
X = add_engineered_features(X_raw)
X.describe().T.head(20)
```

```python
risk_cols = ["LIMIT_BAL", "AGE", "DELINQUENCY_COUNT", "MAX_DELAY", "AVG_DELAY", "UTILIZATION_MEAN", "PAY_TO_BILL_RATIO_MEAN"]
eda = X[risk_cols].copy()
eda[target_col] = y
eda.groupby(target_col)[risk_cols].mean().T
```

```python
plt.figure(figsize=(7, 5))
eda.groupby(target_col)["DELINQUENCY_COUNT"].mean().plot(kind="bar")
plt.title("Average Delinquency Count by Target")
plt.xlabel("Default Next Month")
plt.ylabel("Average Delinquency Count")
plt.tight_layout()
```
